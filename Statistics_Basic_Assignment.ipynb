1) What is the difference between descriptive statistics and inferential
statistics? Explain with examples.
-> Descriptive Statistics - Descriptive statistics refers to the process of summarizing and analyzing data to describe its main features in a clear and meaningful way. It is used to present raw data in a form that makes it easier to understand and interpret. Descriptive statistics involves both graphical representations (such as charts and plots) and numerical measures to summarize data effectively.

Inferential Statistics - Inferential statistics involves using data from a sample to make predictions, generalizations, or conclusions about a larger population. Unlike descriptive statistics, which simply summarizes known data, inferential statistics makes inferences or draws conclusions that go beyond the available data.

2) What is sampling in statistics? Explain the differences between random
and stratified sampling.
-> Sampling in statistics is the process of selecting a subset of individuals or observations from a larger population to make inferences about the whole group. It aims to represent the population without surveying everyone. Random sampling ensures every member has an equal chance of selection, while stratified sampling divides the population into subgroups (strata) based on shared characteristics before random sampling.

Difference
* In random sampling (specifically, simple random sampling), participants are chosen entirely by chance. In stratified sampling, the population is first divided into homogeneous subgroups (strata), and a random sample is taken from each, ensuring every group is represented.
* Random sampling might miss smaller subgroups by chance. Stratified sampling guarantees that all relevant, pre-defined, and non-overlapping subgroups are included proportionally.

3) Define mean, median, and mode. Explain why these measures of central
tendency are important.
-> Mean (Average): The numerical average, calculated by adding all data points and dividing by the total number of points, e.g., for 3, 5, 7, the mean is (3+5+7)/3 = 5.
Median (Middle): The middle value when data is sorted in ascending order. If there is an even number of observations, it is the average of the two middle numbers.
Mode (Most Frequent): The value(s) that appear most frequently in a dataset. A dataset can have one, more than one (bimodal/multimodal), or no mode.

Importance of Measures of Central Tendency
* Data Summarization: They condense complex data into a single, understandable value, making it easier to grasp the "typical" value.
* Handling Outliers (Robustness): The mean is highly sensitive to outliers, while the median is more robust (less affected by extreme, skewed values).
* Identifying Trends: They help identify the center point of data distribution.
* Comparative Analysis: They allow for comparing different datasets (e.g., comparing average income across different regions).

4) Explain skewness and kurtosis. What does a positive skew imply about the
data?
-> Skewness and kurtosis are statistical measures describing the shape of a probability distribution beyond its center and spread. Skewness measures the asymmetry (left or right tail), while kurtosis measures the "tailedness" or peakness of the distribution. Positive skew means the distribution has a longer tail on the right side.
Skewness measures the deviation of a distribution from symmetry.
* Positive Skew (Right-Skewed): The tail on the right side is longer or fatter, and the bulk of the data is concentrated on the left.

```python
#Implement a Python program to compute the mean, median, and mode of a given list of numbers.
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
import statistics
mean_val = statistics.mean(numbers)
median_val = statistics.median(numbers)
mode_vals = statistics.multimode(numbers)
print(f"Mean: {mean_val}")
print(f"Median: {median_val}")
print(f"Mode: {mode_vals}")
```

```python
#Compute the covariance and correlation coefficient between the following two datasets provided as lists in Python:
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
cov_val = np.cov(list_x, list_y)[0, 1]
corr_val = np.corrcoef(list_x, list_y)[0, 1]
print(cov_val)
print(corr_val)
```
